# ***Filter***

```python
def fil(a):
  if a > 5:
    return a
```

```python
out = filter(fil, (1, 4, 7, 3, 9))
```

```python
print(list(out))
```

```python
y = filter(lambda a: a > 5, (1, 4, 7, 3, 9))
```

```python
print(list(y))
```

```python
def veggies_with_e(x):
  return 'e' in x
```

```python
map_obj = filter(veggies_with_e, ['peas', 'garlic', 'ginger', 'beetroot', 'carrot'])
```

```python
print(list(map_obj))
```

```python
map_obj = filter(lambda x: 'r' in x, ['peas', 'garlic', 'ginger', 'beetroot', 'carrot'])
```

```python
print(list(map_obj))
```

# ***Reduce***

```python
from functools import reduce
```

```python
reduce(lambda a, b: a+b, [10, 35, 50, 20])
```

```python
reduce(lambda a, b: a*b, [9, 7, 23, 4])
```

```python
numbers = [3, 4, 6, 9, 34, 12]
```

```python
def custom_sum(first, second):
  return first + second
```

```python
result = reduce(custom_sum, numbers)
```

```python
print(result)
```
